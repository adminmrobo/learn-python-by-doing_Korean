## 0-3 실습 파일 올리기

```python
# 실습 파일 묶음(zip)을 올리고 푼다. 실행하면 아래에 [파일 선택] 버튼이 나온다
from google.colab import files
import zipfile, os
up = files.upload()
zipfile.ZipFile(list(up)[0]).extractall()
print(sorted(f for f in os.listdir() if f.endswith((".csv", ".xlsx", ".json", ".txt"))))
```

```python
# 메뉴판(CSV)을 표로 읽는다. 한글이 섞일 수 있는 CSV 는 utf-8-sig 로 연다
import pandas as pd
menu = pd.read_csv("koi_menu.csv", encoding="utf-8-sig")
print(menu)
```

## 0-4 이것만 알면 코드를 읽는다

```python
shop = "Cafe Koi"                      # 이름 = 값
print("store:", shop)                  # 괄호 안을 화면에 보여 준다
price = {"Americano": 4500, "Latte": 5000}   # 사전: 짝을 지어 둔 표
print("latte:", price["Latte"])
print(menu["price"])                   # 표에서 열 하나 꺼내기
print(menu[menu["price"] >= 5000])     # 조건에 맞는 줄만
print("mean price:", menu["price"].mean())   # 값.함수()
```

## 0-5 실행 규칙 세 가지
일부러 오류를 만나 본다. 맨 아래 줄을 읽고 `"price"` 로 고쳐 다시 실행한다.

```python
print(menu["Price"])
```

## 0-6 한글이 깨질 때

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(menu["menu"], menu["price"], color="#028090")
ax.set(title="Price by menu", xlabel="Menu", ylabel="Price (KRW)")   # 라벨은 영어로
plt.show()
```
